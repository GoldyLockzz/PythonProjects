# 📂 Automated File Organizer (No BS Edition)

## What the hell is this?
This notebook contains a simple Python script that **automatically sorts files inside a folder** into subfolders based on file type/extension.

Basically: instead of manually dragging files around like it’s 1999, this script does the lifting for you.  
## What it does
Given a folder path, the script will:
- Scan every item inside that folder
- **Ignore folders** (so it doesn’t freak out and quit) <br>
- Create subfolders if they don’t exist:<br>
  - `excel/` → `.xlsx`, `.xls`, `.csv`
  - `pdf/` → `.pdf`
  - `images/` → `.jpg`, `.jpeg`, `.png`, `.gif`, `.webp`
  - `others/` → anything else (random crap like `.txt`, `.docx`, `.zip`, etc.)
- Move each file into its correct folder
- Skip files that already exist in the destination folder


## Notes / Disclaimer
- This script **moves** files (not copies). So it will actually change your folder structure.
- If you're paranoid (fair), test it on a dummy folder first.
- If two files have the same name, the script will skip the second one (to avoid overwriting).
- You can easily add more rules/extensions by editing the `RULES` dictionary.
t’s not rocket science… but it’s damn useful.


```python
from pathlib import Path
import shutil

# ====== CONFIG ======
BASE_DIR = Path(r"D:\ONLINE Learning\FreeCodeCamp")

# Map: folder name -> list of extensions
RULES = {
    "excel": [".xlsx", ".xls", ".csv"],
    "pdf": [".pdf"],
    "images": [".jpg", ".jpeg", ".png", ".gif", ".webp"],
}

# Optional: where unknown files go
UNKNOWN_FOLDER = "others"


def organize_files(base_dir: Path) -> None:
    """
    Sort files in base_dir into subfolders based on file extension.
    """
    if not base_dir.exists():
        raise FileNotFoundError(f"Folder not found: {base_dir}")

    # Create destination folders
    for folder in list(RULES.keys()) + [UNKNOWN_FOLDER]:
        (base_dir / folder).mkdir(exist_ok=True)

    moved_count = 0
    skipped_count = 0

    for item in base_dir.iterdir():
        # Skip folders (we only sort files)
        if item.is_dir():
            continue

        ext = item.suffix.lower()
        dest_folder = None

        # Decide destination folder
        for folder, extensions in RULES.items():
            if ext in extensions:
                dest_folder = folder
                break

        if dest_folder is None:
            dest_folder = UNKNOWN_FOLDER

        destination = base_dir / dest_folder / item.name

        # If file already exists there, skip it
        if destination.exists():
            skipped_count += 1
            continue

        shutil.move(str(item), str(destination))
        moved_count += 1

    print(f"✅ Done. Moved: {moved_count} files | Skipped (already existed): {skipped_count} files")


if __name__ == "__main__":
    organize_files(BASE_DIR)
```
